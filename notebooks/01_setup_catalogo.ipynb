# Configuração do Ambiente — Projeto Titanic

## Objetivo

Preparar a estrutura de armazenamento no Databricks para implementar a Arquitetura Medallion no projeto Titanic.

## Etapas

1. Criar o catálogo `projeto_titanic`.
2. Criar os schemas `bronze`, `silver` e `gold`.
3. Criar o volume `projeto_titanic.bronze.raw_data` para armazenar os arquivos brutos.

## Resultado esperado

Disponibilizar a estrutura necessária para a ingestão, transformação, armazenamento e análise dos dados nas camadas Bronze, Silver e Gold.

```python
%sql
CREATE CATALOG IF NOT EXISTS projeto_titanic
COMMENT 'Projeto de portfólio para demonstração de Engenharia e Análise de Dados utilizando Databricks, PySpark, Delta Lake e Arquitetura Medallion com o dataset Titanic.';

CREATE SCHEMA IF NOT EXISTS projeto_titanic.bronze
COMMENT 'Camada Bronze: dados brutos do dataset Titanic, preservados com o mínimo de transformações após a ingestão.';

CREATE SCHEMA IF NOT EXISTS projeto_titanic.silver
COMMENT 'Camada Silver: dados tratados, padronizados e validados, preparados para análises e transformações posteriores.';

CREATE SCHEMA IF NOT EXISTS projeto_titanic.gold
COMMENT 'Camada Gold: dados agregados e estruturados para análises, métricas, indicadores e geração de insights.';

CREATE VOLUME IF NOT EXISTS projeto_titanic.bronze.raw_data
COMMENT 'Armazenamento dos arquivos brutos utilizados na ingestão do projeto Titanic.';
```
